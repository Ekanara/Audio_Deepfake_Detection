# Extract Image

```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn
import cv2 
import os
import shutil
```

# EFS Extract

```python
root_path = "D:\Github\Deepfake_Detection\data\train\efs"
print(os.path.exists(root_path))  # Should be True
print(os.path.abspath(root_path)) # See the full absolute path
```

```python
image_extensions = {".jpg", ".jpeg", ".png", "gif"}
counter = 1
temp_folder=[]
for model_folder in os.listdir(root_path):
    model_path = os.path.join(root_path, model_folder)
    if not os.path.isdir(model_path):
        continue

    for dirpath, _, filenames in os.walk(model_path, topdown=False):
        if dirpath == model_path:
            continue  # skip the model folder itself

        for filename in filenames:
            ext = os.path.splitext(filename)[1].lower()
            if ext in image_extensions:
                src = os.path.join(dirpath, filename)
                dst = os.path.join(model_path, filename)

                # Handle duplicates by adding (1), (2), etc.
                base, ext = os.path.splitext(dst)
                counter = 1
                while os.path.exists(dst):
                    dst = f"{base}({counter}){ext}"
                    counter += 1

                shutil.move(src, dst)

        # Delete the folder after moving images
        try:
            os.rmdir(dirpath)
        except OSError:
            pass

print("All images moved up one level and subfolders deleted.")
```

# FR Extract

```python
root_path = r"D:\Github\Deepfake_Detection\data\\train\real"

for folder in os.listdir(root_path):
    folder_path = os.path.join(root_path, folder)
    if os.path.isdir(folder_path):
        count = 0
        for root, _, files in os.walk(folder_path):
            if root == folder_path:  # Skip the top level, we want to move subfolder files
                continue
            for f in files:
                if os.path.splitext(f)[1].lower() in image_extensions:
                    src_path = os.path.join(root, f)
                    dst_path = os.path.join(folder_path, f)

                    # Avoid overwriting duplicates
                    if os.path.exists(dst_path):
                        name, ext = os.path.splitext(f)
                        i = 1
                        while os.path.exists(dst_path):
                            dst_path = os.path.join(folder_path, f"{name}_{i}{ext}")
                            i += 1

                    shutil.move(src_path, dst_path)  # change to move if needed
                    count += 1

        print(f"{folder}: Extracted {count} images directly into {folder_path}")
```

# FS Extract

```python
root_path = r"D:\Github\Deepfake_Detection\data\train\fs"  # The "fr" folder

for model_folder in os.listdir(root_path):
    model_path = os.path.join(root_path, model_folder)
    if not os.path.isdir(model_path):
        continue

    frames_path = os.path.join(model_path, "frames")
    if not os.path.isdir(frames_path):
        continue

    # Move all images from inside 'frames' to the model's main folder
    for dirpath, _, filenames in os.walk(frames_path, topdown=False):
        for filename in filenames:
            ext = os.path.splitext(filename)[1].lower()
            if ext in image_extensions:
                src = os.path.join(dirpath, filename)
                dst = os.path.join(model_path, filename)

                # Handle duplicates
                base, ext = os.path.splitext(dst)
                counter = 1
                while os.path.exists(dst):
                    dst = f"{base}({counter}){ext}"
                    counter += 1

                shutil.move(src, dst)

    # Remove the 'frames' folder after moving images
    shutil.rmtree(frames_path, ignore_errors=True)

print("Images moved from 'frames' into model folders, and 'frames' deleted.")
```
